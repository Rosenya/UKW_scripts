## 2. Czym są języki skryptowe?

Języki skryptowe to języki wysokiego poziomu, które:
- są interpretowane (kod wykonywany jest przez interpreter),
- mają uproszczoną składnię,
- pozwalają szybko tworzyć działające programy,
- są szeroko wykorzystywane w analizie danych, automatyzacji, web-development oraz AI.

Python jest jednym z najpopularniejszych języków skryptowych na świecie dzięki czytelnej składni, dużej liczbie bibliotek, wszechstronności zastosowań.

## 8. Formatowanie tekstu

- Łączenie tekstu (konkatenacja)

```python
imie = "Anna"
print("Cześć " + imie)
```

- Powielanie tekstu

```python
print("Ha" * 3)  
```

- Długość tekstu

```python
tekst = "Python"
print(len(tekst)) 
```

- f-string

```python
imie = "Jan"
wiek = 20

print(f"Cześć {imie}, masz {wiek} lat.")
```

takie nawiasy klamrowe mogą również zawierać obliczenia:

```python
print(f"Za 5 lat będziesz mieć {wiek + 5} lat.")
```

- funkcja .upper() – zamiana na wielkie litery

```python
tekst = "python"
print(tekst.upper()) 
```

- funkcja .lower() – zamiana na małe litery

```python
tekst = "PYTHON"
print(tekst.lower())  
```

- funkcja .capitalize() – pierwsza litera wielka

```python
tekst = "python jest super"
print(tekst.capitalize())
```

- funkcja .title() – każda pierwsza litera wyrazu wielka

```python
tekst = "programowanie w pythonie"
print(tekst.title())
```

- funkcja .strip() – usuwanie białych znaków

```python
tekst = "   Python   "
print(tekst.strip())
```

- funkcja .replace() – zamiana fragmentu tekstu

```python
tekst = "Lubię Java"
print(tekst.replace("Java", "Python"))
```

- funkcja .split() – podział tekstu

```python
tekst = "Ala ma kota"
print(tekst.split())
```

- Takie funkcje można ze sobą łączyć:

```python
tekst = "Ala ma kota"
tekst.strip().lower().replace("a", "o")
```

## Zadanie 25
Użyj funkcji: .strip(), .isalnum(), .isalpha(), .isdigit()
Wczytaj od użytkownika:
- imię,
- nazwisko,
- adres e-mail.

Następnie:
- usuń zbędne spacje z początku i końca każdego pola,
- zapisz imię i nazwisko z wielkiej litery,
- zapisz adres e-mail małymi literami,
- wyświetl komunikat w formacie:
- Użytkownik: Jan Kowalski
- Email: jan.kowalski@gmail.com

```python
imie = input("Podaj imię: ")
nazwisko = input("Podaj nazwisko: ")
email = input("Podaj adres e-mail: ")

imie = imie.strip()
nazwisko = nazwisko.strip()
email = email.strip()

pelne_nazwisko = imie.capitalize() + " " + nazwisko.capitalize()
email = email.lower()

print(f"Użytkownik: {pelne_nazwisko}, \nE-mail: {email}")

```
