# Funkcje

Funkcja to blok kodu, który wykonuje określone zadanie i może być wielokrotnie używany.

Dlaczego używamy funkcji?
- zmniejszają powtarzalność kodu
- zwiększają czytelność
- ułatwiają testowanie
- dzielą program na logiczne części

Składnia:

```python
def nazwa_funkcji():
    instrukcje
```

Przykład:

```python
def przywitaj():
    print("Witaj w programie!")

przywitaj()
```

## Parametry funkcji
Parametry pozwalają przekazywać dane do funkcji.

```python
def przywitaj(imie):
    print("Cześć", imie)

przywitaj("Anna")
```

Dla wielu parametrów:

```python
def suma(a, b):
    print(a + b)

suma(5, 3)
```

## Return

```python
def suma(a, b):
    return a + b

wynik = suma(5, 3)
print(wynik)
```

Dla wielu wartości:

```python
def dane():
    return "Anna", 21

imie, wiek = dane()
print(imie, wiek)
```

## Zagnieżdżenie funkcji

```python
def wiadomosc(a):
    print("X = ",a)

def pomnoz_2(x):
    x=x*2
    print("X = ",x)
    return x

x=10

wiadomosc(x)
pomnoz_2(x)
wiadomosc(x)
```

## REKURENCJA

```python
def nieUjemna(a):
    if a>0:
        print(a," jest nie ujemna")
        a=a-1
        return nieUjemna(a)
    else:
        print("a jest rowne 0")

nieUjemna(10)
```

## Parametry domyślne

```python
def przywitaj(imie="Gość"):
    print("Cześć", imie)

przywitaj()
przywitaj("Jan")
```

## Argumenty nazwane

```python
def dane(imie, wiek):
    print(imie, wiek)

dane(wiek=20, imie="Anna")
```

## Zakres zmiennych

Zmienna lokalna:

```python
def test():
    x = 5
    print(x)

test()
```

Zmienna globalna:

```python
x = 10

def test():
    print(x)

test()
```

Modyfikacja zmiennej globalnej:

```python
x = 10

def zmien():
    global x
    x = 20

zmien()
print(x)
```

1. Przyjmuje listę **30 liczb całkowitych**,

3. Usuwa duplikaty,

5. Oblicza:

   * minimum,

# Zadanie 2 
Napisz funkcję `analiza_studentow(studenci)`, która:

1. Przyjmuje listę słowników studentów,
2. Tworzy listę studentów z oceną ≥ 4,
3. Sortuje ich malejąco według oceny,
4. Zamienia imiona na wielkie litery,
5. Oblicza średnią ocen,
6. Zwraca:

   * listę najlepszych studentów,
   * średnią ocen.


```python
studenci = [
{"imie": "anna", "wiek": 21, "ocena": 5},
{"imie": "jan", "wiek": 22, "ocena": 3},
{"imie": "maria", "wiek": 20, "ocena": 4},
{"imie": "piotr", "wiek": 23, "ocena": 2},
{"imie": "ola", "wiek": 21, "ocena": 5},
{"imie": "tomek", "wiek": 24, "ocena": 4}
]
```

```python

studenci = [
{"imie": "anna", "wiek": 21, "ocena": 5},
{"imie": "jan", "wiek": 22, "ocena": 3},
{"imie": "maria", "wiek": 20, "ocena": 4},
{"imie": "piotr", "wiek": 23, "ocena": 2},
{"imie": "ola", "wiek": 21, "ocena": 5},
{"imie": "tomek", "wiek": 24, "ocena": 4}
]
def analiza_studentow(studenci):
    najlepsi = [s for s in studenci if s["ocena"] >= 4]
    najlepsi.sort(
        key=lambda x: x["ocena"],
        reverse=True
    )
    for s in najlepsi:
        s["imie"] = s["imie"].upper()
    srednia = sum(
        s["ocena"] for s in studenci
    ) / len(studenci)
    return najlepsi, srednia

print(analiza_studentow(studenci))

```

# Zadanie 3 
Napisz funkcję `analiza_tekstu(tekst)`, która:

1. Zamienia tekst na małe litery,
2. Usuwa znaki interpunkcyjne (`. , ! ?`),
3. Dzieli tekst na słowa,
4. Zlicza wystąpienia każdego słowa,
5. Znajduje najczęściej występujące słowo,
6. Zwraca słownik wyników.


```python
tekst = """
Python jest szybki. Python jest prosty!
Python jest popularny, i Python jest użyteczny.
"""
```

```python
import re
from typing import Counter
tekst = """
Python jest szybki. Python jest prosty!
Python jest popularny, i Python jest użyteczny.
"""
def analiza_tekstu(tekst):
    tekst = tekst.lower()
    for znak in [".", ",", "!", "?"]:
        tekst = tekst.replace(znak, "")
    slowa = tekst.split()
    liczba_slow = len(slowa)
    counter = Counter(slowa)
    najczestsze = counter.most_common(1)
    unikalne_slowa = set(slowa)
    liczba_unikalnych_slow = len(unikalne_slowa)

    return liczba_slow, counter, najczestsze, liczba_unikalnych_slow, slowa
print(analiza_tekstu(tekst))
```

1. Przyjmuje macierz **5×6 (30 elementów)**,
2. Oblicza sumę każdego wiersza,
3. Znajduje największy element macierzy,
4. Spłaszcza macierz do jednej listy,
5. Sortuje wynik rosnąco,
6. Zwraca wyniki w słowniku.


```python
macierz = [
[5, 12, 7, 9, 3, 15],
[20, 1, 14, 8, 6, 2],
[11, 18, 4, 10, 16, 13],
[25, 17, 19, 21, 22, 23],
[24, 26, 27, 28, 29, 30]
]
```

```python
macierz = [
[5, 12, 7, 9, 3, 15],
[20, 1, 14, 8, 6, 2],
[11, 18, 4, 10, 16, 13],
[25, 17, 19, 21, 22, 23],
[24, 26, 27, 28, 29, 30]
]
def analiza_macierzy(macierz):
    suma_wierszy = [sum(wiersz) for wiersz in macierz]
    plaska_macierz = [element for wiersz in macierz for element in wiersz]
    najwiekszy_element = max(plaska_macierz)
    posortowana_lista = sorted(plaska_macierz)
    return {
      "sumy_wierszy": suma_wierszy,
      "maksimum": najwiekszy_element,
      "plaska_posortowana": posortowana_lista,
  }
print(analiza_macierzy(macierz))
```
