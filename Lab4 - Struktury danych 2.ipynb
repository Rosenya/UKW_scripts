# Struktury danych cz. II

## 1. Zagnieżdżone struktury danych

W Pythonie struktury danych mogą zawierać inne struktury danych.

### Lista list - macierz

```python
macierz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(macierz[0][1])  
```

### Lista słowników

```python
studenci = [
    {"imie": "Anna", "ocena": 5},
    {"imie": "Jan", "ocena": 4}
]

print(studenci[0]["imie"])
```

### Słownik list

```python
dane = {
    "liczby": [1, 2, 3],
    "imiona": ["Anna", "Jan"]
}

print(dane["liczby"][1])
```

### Iteracja po zagnieżdżonych strukturach

```python
studenci = [
    {"imie": "Anna", "ocena": 5},
    {"imie": "Jan", "ocena": 4}
]

for student in studenci:
    print(student["imie"], student["ocena"])
```

## 2. Sortowanie

### Sortowanie listy

```python
liczby = [5, 2, 9, 1]
liczby.sort()
print(liczby)
```

### Sortowanie malejące

```python
liczby.sort(reverse=True)
```

### Stworzenie nowej listy posortowanej

```python
lista = [3,5,2,9]
lista2= sorted(lista)
print(lista)
print(lista2)
```

### Sortowanie w słownikach

Konieczne jest by klucze były takie same w całym słowniku a wartości numeryczne.

```python
dane = [{"A":7,"B":5},
{"A":3,"B":3},
{"A":1,"B":6},
{"A":1,"B":10},
{"A":4,"B":4}]

dane.sort(key=lambda x: x["A"])
print(dane)
```

## Filtrowanie danych

### Klasyczne filtrowanie

```python
liczby = [1,2,3,4,5,6]

parzyste = []

for x in liczby:
    if x % 2 == 0:
        parzyste.append(x)

print(parzyste)
```

## Wyszukiwanie

### Sprawdzenie czy element istnieje

```python
if 3 in [1,2,3]:
    print("Jest")
```

### Wyszukiwanie minimum/maksimum

```python
print(min([1,5,2]))
print(max([1,5,2]))
```

### Wyszukiwanie w słowniku

```python
student = {"imie": "Anna","wiek":20}

if "imie" in student:
    print(student["wiek"])
```
