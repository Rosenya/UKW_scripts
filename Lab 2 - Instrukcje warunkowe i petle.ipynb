# Instrukcje warunkowe; Pętle

## 1. Instrukcje warunkowe 
Instrukcje warunkowe pozwalają programowi podejmować decyzje.

```python
if warunek:
    # kod
elif inny_warunek:
    # kod
else:
    # kod
```

Przykład 1 – Pełnoletność

```python
wiek = int(input("Podaj wiek: "))

if wiek >= 18:
    print("Pełnoletni")
else:
    print("Niepełnoletni")
```

- Operatory logiczne

not

```python
x = 10

if x > 5 and x < 20:
    print("Liczba w zakresie (5,20)")
```

## 2. Pętle

Pętle pozwalają powtarzać kod.

- Pętla for

Używana, gdy znamy liczbę powtórzeń.

```python
for i in range(5):
    print(i)
```

Funkcja range()

range(start, stop, krok)

- Pętla while

Używana, gdy liczba powtórzeń zależy od warunku. Jest wykonywana tak długo jak warunek jest prawdziwy

```python
x = 0

while x < 5:
    print(x)
    x += 1
```

# Zadanie 17 (trudne)
Policz samogłoski we wpisanym przez użytkownika słowie

```python
tekst = input("Wpisz tekst: ").lower()
vowels = "aeiouy"
vowel_count = 0
for char in tekst:
    if char in vowels:
        vowel_count += 1
print("Liczba samogłosek w tekście:", tekst, "to", vowel_count)
```

# Zadanie 23 (trudne)
Sprawdź czy liczba jest pierwsza

```python
liczba = int(input("Wpisz liczbę: "))
czy_pierwsza = True
if liczba <= 1:
    czy_pierwsza = False
else:
    for i in range(2, int(liczba**0.5) + 1):
        if liczba % i == 0:
            czy_pierwsza = False
            break
if czy_pierwsza:
    print(f"Liczba {liczba} jest liczbą pierwszą!")
else:
    print(f"Liczba {liczba} NIE jest liczbą pierwszą.")
```

# Zadanie 24
FizzBuzz (1–30)
Wypisuj liczby od 1 do 30. Dla liczb podzielnych przez 3 wyświetl "Fizz", dla liczb podzielnych przez 5 - "Buzz", a dla liczb podzielnych przez 5 oraz 3 wypisz "FizzBuzz".

```python
for i in range(1, 31):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```
