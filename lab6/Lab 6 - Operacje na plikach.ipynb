```python
plik = open("nazwa pliku", "tryb otwierania")
```

Tryby:
- "r" - odczyt

Zalecana metoda, która automatycznie zamyka plik po odczycie:

```python
with open("dane.txt", "r") as plik:
    tekst = plik.read()

print(tekst)
```

CSV (Comma-Separated Values) to prosty format pliku tekstowego służący do przechowywania danych tabelarycznych. Każdy wiersz pliku odpowiada wierszowi tabeli, a dane w komórkach są rozdzielone przecinkami. Pliki .csv są powszechnie używane do wymiany danych między arkuszami kalkulacyjnymi (np. Excel)

Zapisz poniższe dane w formacie .csv, możesz to zrobić za pomocą notatnika

- odczyt jako lista

```python
import csv

with open("studenci.csv", "r") as plik:

    reader = csv.reader(plik)

    for wiersz in reader:
        print(wiersz)
with open("studenci.csv", "w") as plik:
    writer = csv.writer(plik)
    writer.writerow(["Imie", "Wiek", "Ocena"])
    writer.writerow(["Anna", 21, 5])
    writer.writerow(["Jan", 22, 3])
    writer.writerow(["Maria", 20, 4])
    writer.writerow(["Piotr", 23, 2])
    writer.writerow(["Ola", 21, 5])
```

# Wyzwanie 1 — Analiza listy liczb z pliku

Plik wejściowy: liczby.txt


```python
12
45
78
23
56
89
34
67
90
11
22
33
44
55
66
77
88
99
100
5
17
28
39
40
51
62
73
84
95
10
```

Napisz program, który:

Wczyta liczby z pliku liczby.txt.
Zapisze je do listy liczb całkowitych.

Utworzy funkcje:
- funkcja filtruj_liczby(lista), która zwraca liczby większe niż 50
- funkcja statystyki(lista), która zwraca: minimum, maksimum, średnią

Zapisze wynik do pliku: wynik1.txt


```python
def filtruj_liczby(lista):
    return [x for x in lista if x > 50]


def statystyki(lista):
    minimum = min(lista)
    maksimum = max(lista)
    srednia = sum(lista) / len(lista)
    return minimum, maksimum, srednia

with open("liczby.txt", "w") as plik:
    plik.write(
        "12\n45\n78\n23\n56\n89\n34\n67\n90\n11\n"
        "22\n33\n44\n55\n66\n77\n88\n99\n100\n5\n"
        "17\n28\n39\n40\n51\n62\n73\n84\n95\n10\n"
    )

with open("liczby.txt", "r") as plik:
    liczby = [int(linia.strip()) for linia in plik if linia.strip()]

wieksze_niz_50 = filtruj_liczby(liczby)
mini, maksi, srednia = statystyki(liczby)

with open("wynik1.txt", "w") as plik:
    plik.write(f"Liczby wieksze niz 50: {wieksze_niz_50}\n")
    plik.write(f"Minimum: {mini}\n")
    plik.write(f"Maksimum: {maksi}\n")
    plik.write(f"Srednia: {srednia}\n")
```

# Wyzwanie 2 — System ocen studentów
Plik wejściowy: studenci.txt


```python
Anna,5
Jan,3
Maria,4
Piotr,2
Ola,5
Tomek,4
Karol,3
Ewa,5
Adam,2
Natalia,4
```

Napisz program, który:

Wczyta dane studentów z pliku.
Zapisze je jako listę słowników:
{"imie": "Anna", "ocena": 5}

Utworzy funkcje:
- funkcja najlepsi(studenci), która zwraca studentów z oceną ≥ 4
- funkcja srednia_ocen(studenci), która oblicza średnią ocen

Zapisze wynik do pliku: raport_studentow.txt


```python
with open("studenci.txt", "w") as plik:
    plik.write(
        "Anna,5\n"
        "Jan,3\n"
        "Maria,4\n"
        "Piotr,2\n"
        "Ola,5\n"
        "Tomek,4\n"
        "Karol,3\n"
        "Ewa,5\n"
        "Adam,2\n"
        "Natalia,4\n"
    )


def najlepsi(studenci):
    return [s for s in studenci if s["ocena"] >= 4]


def srednia_ocen(studenci):
    return sum(s["ocena"] for s in studenci) / len(studenci)


studenci = []
with open("studenci.txt", "r") as plik:
    for linia in plik:
        if linia.strip():
            imie, ocena = linia.strip().split(",")
            studenci.append({"imie": imie, "ocena": int(ocena)})

top_studenci = najlepsi(studenci)
srednia = srednia_ocen(studenci)

with open("raport_studentow.txt", "w") as plik:
    plik.write(f"Srednia ocen: {srednia}\n")
    plik.write("Najlepsi studenci:\n")
    for s in top_studenci:
        plik.write(f"{s['imie']}: {s['ocena']}\n")
```

# Wyzwanie 3 — Analiza sprzedaży
Plik wejściowy: sprzedaz.txt


```python
Laptop,3000
Mysz,50
Klawiatura,150
Monitor,800
Drukarka,600
Tablet,1200
Słuchawki,200
Telefon,2500
Kamera,900
Router,180
```

Napisz program, który:

Wczyta produkty do słownika:
{"Laptop": 3000}

Utworzy funkcje:
- funkcja produkty_drogie(produkty), która zwraca produkty droższe niż 500
- funkcja promocja(produkty), która obniża ceny o 10%

Zapisze nowe ceny do pliku: promocja.txt


```python
with open("sprzedaz.txt", "w") as plik:
    plik.write(
        "Laptop,3000\n"
        "Mysz,50\n"
        "Klawiatura,150\n"
        "Monitor,800\n"
        "Drukarka,600\n"
        "Tablet,1200\n"
        "Słuchawki,200\n"
        "Telefon,2500\n"
        "Kamera,900\n"
        "Router,180\n"
    )


def produkty_drogie(produkty):
    return {nazwa: cena for nazwa, cena in produkty.items() if cena > 500}


def promocja(produkty):
    return {nazwa: cena * 0.9 for nazwa, cena in produkty.items()}


produkty = {}
with open("sprzedaz.txt", "r") as plik:
    for linia in plik:
        if linia.strip():
            nazwa, cena = linia.strip().split(",")
            produkty[nazwa] = float(cena)

drogie = produkty_drogie(produkty)
po_promocji = promocja(produkty)

print(drogie)

with open("promocja.txt", "w") as plik:
    for nazwa, nowa_cena in po_promocji.items():
        plik.write(f"{nazwa},{nowa_cena}\n")
```

# Wyzwanie 4 — Liczenie słów w tekście
Plik wejściowy: tekst.txt


```python
Python jest prosty.
Python jest szybki.
Programowanie w Python jest przyjemne.
Python jest popularny.
```

Napisz program, który:
- Wczyta tekst.
- Zamieni litery na małe.
- Usunie znaki: . , !
- Policzy ile razy występuje każde słowo.

Zapisze wynik do pliku: statystyki_slow.txt


```python
with open("tekst.txt", "w") as plik:
    plik.write(
        "Python jest prosty.\n"
        "Python jest szybki.\n"
        "Programowanie w Python jest przyjemne.\n"
        "Python jest popularny.\n"
    )


with open("tekst.txt", "r") as plik:
    tekst = plik.read().lower()

for znak in [".", ",", "!"]:
    tekst = tekst.replace(znak, "")

slowa = tekst.split()
licznik = {}

for slowo in slowa:
    licznik[slowo] = licznik.get(slowo, 0) + 1

with open("statystyki_slow.txt", "w") as plik:
    for slowo, ilosc in licznik.items():
        plik.write(f"{slowo}: {ilosc}\n")
```
