# Wstęp do programowania obiektowego

Programowanie obiektowe (OOP – Object-Oriented Programming) polega na tworzeniu klas i obiektów, łączeniu danych i funkcji, modelowaniu rzeczywistych obiektów.

## Klasa

Klasa to szablon obiektu.

```python
class Samochod:
    pass #pass oznacza pustą klasę

auto = Samochod() # Tworzenie obiektu
print(auto)
```

## Atrybuty

Atrybuty to zmienne w klasie.

```python
class Samochod:

    marka = "Toyota"
    kolor = "czerwony"

auto = Samochod() # Tworzenie obiektu

print(auto.marka) # Podgląd atrybutu
```

## Metody

Metody to funkcje wewnątrz klasy.

```python
class Samochod:

    def jedz(self):             # self oznacza aktualny obiekt
        print("Samochód jedzie")

auto = Samochod()
auto.jedz()
```

## Metody z parametrami


```python
class Samochod:
    marka = "Opel"
    def opisz(self):      
        print(f"Samochod marki {self.marka}")

auto = Samochod()
auto.opisz()
```

```python
class Samochod:
    predkosc = 100

    def przyspiesz(self,wartosc):
        return self.predkosc+wartosc

auto = Samochod()

print(f"Samochod jedzie z predkoscia {auto.predkosc}")

nowa_predkosc=auto.przyspiesz(30)

print(f"Samochod jedzie z predkoscia {nowa_predkosc}")
```

## Konstruktor __ init __

Konstruktor uruchamia się przy tworzeniu obiektu.

```python
class Samochod:

    def __init__(self, wpisana_marka):

        self.marka = wpisana_marka


auto = Samochod("Toyota")

print(auto.marka)
```

Konstruktor wykorzystany do nadania wartości domyślnych:

```python
class Samochod:

    def __init__(self):

        self.marka = "nieznana"
    
    def przypisz_marke(self,podana_marka):
        self.marka=podana_marka

    def opisz(self):
        print(f"Marka samochodu to {self.marka}")


auto = Samochod()

auto.opisz()

auto.przypisz_marke("Opel")

auto.opisz()
```

## Dziedziczenie
Dziedziczenie pozwala tworzyć nową klasę na podstawie istniejącej klasy.

Nowa klasa: dziedziczy atrybuty, metody, może dodawać nowe elementy i może zmieniać istniejące metody.

```python
class Osoba:

    def __init__(self, imie):
        self.imie = imie

    def przedstaw_sie(self):
        print("Mam na imię", self.imie)


class Student(Osoba):

    def __init__(self, imie, kierunek):
        super().__init__(imie)  # super() pozwala wywołać konstruktor klasy bazowej
        self.kierunek = kierunek


s = Student("Anna", "Informatyka")

s.przedstaw_sie()
print(s.kierunek)
```

### Nadpisywanie metod 

```python
class Student(Osoba):

    def przedstaw_sie(self):

        print("Jestem studentem.")
```

Ta metoda zastępuje metodę z klasy bazowej.

## Polimorfizm
Polimorfizm oznacza że ta sama metoda może działać różnie dla różnych obiektów

```python
class Pies:

    def dzwiek(self):
        print("Hau hau")


class Kot:

    def dzwiek(self):
        print("Miau")


zwierzeta = [Pies(), Kot()]

for z in zwierzeta:

    z.dzwiek()
```

## Enkapsulacja 

Polega na ukrywaniu danych.

```python
class Konto:
    saldo=1000

    def stan_konta(self):
        print(f"Stan konta: {self.saldo}")

k1 = Konto()
k1.stan_konta()

k1.saldo+=500

k1.stan_konta()
```

Jak zablokować modyfikację zmiennej saldo?

```python
class Konto:
    __saldo=1000

    def stan_konta(self):
        print(f"Stan konta: {self.__saldo}")

k1 = Konto()
k1.stan_konta()

k1.__saldo+=500

k1.stan_konta()
```

Podwójne _ oznacza prywatny atrybut.

```python
class Konto:
    __saldo=1000

    def stan_konta(self):
        print(f"Stan konta: {self.__saldo}")
    
    def zmien_stan_konta(self,wartosc):
        self.__saldo+=wartosc

k1 = Konto()
k1.stan_konta()

k1.zmien_stan_konta(500)

k1.stan_konta()
```

## Inner classes
To klasa zdefiniowana wewnątrz innej klasy.

Stosuje się ją, gdy jedna klasa jest logicznie związana z drugą oraz gdy chcemy uporządkować kod.

```python
class Uczelnia:

    class Student:

        def __init__(self, imie):

            self.imie = imie

        def pokaz(self):

            print("Student:", self.imie)


student1 = Uczelnia.Student("Anna")

student1.pokaz()
```

### Wyzwanie 1
Utwórz system do obsługi pracowników firmy.

- Utwórz klasę bazową Pracownik, która:
    - posiada konstruktor przyjmujący: , imie, nazwisko, pensja
    - posiada metodę pokaz_dane() — wyświetla dane pracownika

- Utwórz klasę pochodną Manager, która:
    - dziedziczy po Pracownik
    - posiada dodatkowy atrybut dzial
    - nadpisuje metodę pokaz_dane(), dodając informację o dziale

Utwórz listę obiektów: 2 pracowników i 2 managerów.
Wyświetl dane wszystkich pracowników przy użyciu pętli.

```python
class Pracownik:

    def __init__(self, imie, nazwisko, pensja):
        self.imie = imie
        self.nazwisko = nazwisko
        self.pensja = pensja

    def pokaz_dane(self):
        return f"{self.imie} {self.nazwisko}, pensja: {self.pensja}"

class Manager(Pracownik):

    def __init__(self, imie, nazwisko, pensja, dzial):
        super().__init__(imie, nazwisko, pensja)
        self.dzial = dzial

    def pokaz_dane(self):
        return f"{self.imie} {self.nazwisko}, pensja: {self.pensja}, dział: {self.dzial}"

lista_pracownikow = [
    Pracownik("Jan", "Kowalski", 3000),
    Manager("Anna", "Nowak", 5000, "IT"),
    Pracownik("Piotr", "Wiśniewski", 3500),
    Manager("Katarzyna", "Zielińska", 6000, "HR")
]

for pracownik in lista_pracownikow:
    print(pracownik.pokaz_dane())
```

### Wyzwanie 2
Utwórz system obsługi zamówień sklepu.

- Utwórz klasę Produkt, która posiada nazwę oraz cenę jako atrybuty
- Utwórz klasę Zamowienie, która:
    - posiada listę produktów
    - posiada metody: dodaj_produkt(), oblicz_wartosc(), pokaz_zamowienie()

- Dodaj kilka produktów do zamówienia.
- Oblicz całkowitą wartość zamówienia.
- Zapisz zamówienie do pliku zamowienie.txt

```python
class Produkt:

    def __init__(self, nazwa, cena):
        self.nazwa = nazwa
        self.cena = cena

class Zamowienie:
    def __init__(self):
        self.produkty = []

    def dodaj_produkt(self, nazwa, cena, ilosc):
        produkt = Produkt(nazwa, cena)
        produkt.ilosc = ilosc
        self.produkty.append(produkt)

    def oblicz_wartosc(self):
        total = 0
        for produkt in self.produkty:
            total += produkt.cena * produkt.ilosc
        return total

    def pokaz_zamowienie(self):
        for produkt in self.produkty:
            print(f"Produkt: {produkt.nazwa}, Cena: {produkt.cena}, Ilość: {produkt.ilosc}")
        print(f"Wartość zamówienia: {self.oblicz_wartosc()}")

    def zapisz_do_pliku(self, nazwa_pliku="zamowienie.txt"):
        with open(nazwa_pliku, "w", encoding="utf-8") as plik:
            for produkt in self.produkty:
                plik.write(f"Produkt: {produkt.nazwa}, Cena: {produkt.cena}, Ilość: {produkt.ilosc}\n")
            plik.write(f"Wartość zamówienia: {self.oblicz_wartosc()}\n")

zamowienie1 = Zamowienie()
zamowienie1.dodaj_produkt("monitor", 500, 3)
zamowienie1.dodaj_produkt("klawiatura", 100, 1)
zamowienie1.pokaz_zamowienie()
zamowienie1.zapisz_do_pliku("zamowienie.txt")
```
