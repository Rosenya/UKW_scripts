# Struktury danych

W praktycznym programowaniu bardzo rzadko operujemy tylko na pojedynczych zmiennych. Zwykle musimy przechowywać **zbiory danych**: listę liczb, zestaw nazwisk, słownik tłumaczeń itp.

W Pythonie najczęściej używane struktury danych to:

* **lista (`list`)**
* **krotka (`tuple`)**
* **słownik (`dict`)**
* **zbiór (`set`)**

Każda z tych struktur ma inne właściwości i zastosowania.

## 1. Lista

Lista to uporządkowana kolekcja elementów, która może zawierać różne typy danych.

### Tworzenie listy

```python
liczby = [1, 2, 3, 4]
imiona = ["Anna", "Jan", "Maria"]
mieszana = [1, "Python", 3.14]
```

### Uporządkowanie elementów
Indeksy są liczone od 0.

```python
imiona = ["Anna", "Jan", "Maria"]

print(imiona[0])  
print(imiona[2])   
```

### Modyfikacje. Zmiana elementu

```python
liczby = [1, 2, 3]

liczby[0] = 10
print(liczby)
```

### Modyfikacje. Dodawanie elementu

```python
lista = [1, 2, 3]

lista.append(4)
print(lista)
```

### Modyfikacje. Wstawianie elementu

```python
lista = [1, 2, 3]
lista.insert(1, 100)
print(lista)
```

### Modyfikacje. Usuwanie elementu

```python
lista = [1, 2, 3]
lista.remove(3) # usuwa element o wartości 3
print(lista)
```

```python
lista = [1, 2, 3]
lista.pop(-1) # usuwa element o podanym indeksie, domyślnie jest ostatni
print(lista)
```

### Długość listy

```python
lista = [1, 2, 3]
print(len(lista))
```

## 3. Krotki (tuples)

Krotka jest podobna do listy, ale **nie można jej zmieniać**.

```python
punkt = (3, 5)
kolory = ("red", "green", "blue")

print(punkt)
print(kolory)

print(punkt[1])
print(kolory[1])
```

## 4. Słowniki

Słownik przechowuje dane w formie **klucz → wartość**.

```python
student = {
    "imie": "Anna",
    "wiek": 21,
    "kierunek": "Informatyka"
}

print(student)

print(student["wiek"])
```

### Dodawanie pary

```python
student["ocena"] = 5
```

### Iteracja po parach

```python
for k, v in student.items():
    print(k, v)
```

### Iteracja po kluczach

```python
for klucz in student:
    print(klucz, student[klucz])
```

## 5. Zbiory
Zestaw unikalnych elementów.

```python
liczby = {1, 2, 3, 3, 3, 3, 3, 3, 4}
print(liczby)
```

### Dodawanie elementu do zbioru

```python
liczby.add(10)
```

### Operacje na zbiorach

```python
A = {1,2,3}
B = {3,4,5}

print(A | B)   # suma
print(A & B)   # część wspólna
print(A - B)   # różnica
```
